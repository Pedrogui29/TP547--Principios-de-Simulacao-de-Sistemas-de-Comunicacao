```python
import numpy as np


p = 0.3
N = 100000
amostras = []

# a) Simulação para gerar amostras da distribuição geométrica
for _ in range(N):
    insucessos = 0

    while True:

        u = np.random.uniform(0, 1)
        if u <= p:

            break
        else:

            insucessos += 1
    amostras.append(insucessos)


amostras_np = np.array(amostras)

# b) Probabilidade de 3 insucessos
count_b = np.sum(amostras_np == 3)
prob_b = count_b / N

# c) Probabilidade de precisar de n> 5 insucessos
count_c = np.sum(amostras_np > 5)
prob_c = count_c / N

# d) Média e Variância
media_simulada = np.mean(amostras_np)
variancia_simulada = np.var(amostras_np)

print(f"b) Probabilidade (Sucesso na 4ª tentativa): {prob_b:.4f}")
print(f"c) Probabilidade (Mais de 6 tentativas): {prob_c:.4f}")
print(f"d) Média Simulada: {media_simulada:.4f}")
print(f"d) Variância Simulada: {variancia_simulada:.4f}")
```

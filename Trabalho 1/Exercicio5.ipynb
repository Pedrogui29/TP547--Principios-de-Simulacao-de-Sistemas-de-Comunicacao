```python
import numpy as np
import matplotlib.pyplot as plt

N = 1000

# vetor U com distribuição uniforme entre 0 e 1 [2, 5]
u = np.random.uniform(0, 1, N)

# x usando a transformada inversa: x = sqrt(u) [2, 4]
X = np.sqrt(u)

# valores para a linha da PDF teórica f(x) = 2x
x_teorico = np.linspace(0, 1, 100)
pdf_teorica = 2 * x_teorico

plt.figure(figsize=(8, 5))
plt.hist(X, bins=30, density=True, alpha=0.6, label='Simulado (Histograma)')
plt.plot(x_teorico, pdf_teorica, 'r', linewidth=2, label='Teórico (f(x)=2x)')
plt.title('Comparação: Histograma Simulado vs PDF Teórica')
plt.xlabel('x')
plt.ylabel('Densidade')
plt.legend()
plt.show()
```

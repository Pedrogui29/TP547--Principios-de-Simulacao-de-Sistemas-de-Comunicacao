```python
import numpy as np



lambda1=4 #Número médio de requisições
N=1000000 #Numero de amostras
value=2
count_a=0
count_b=0
av=np.array([])
x=np.random.uniform(0,1,N)
for ix in x:
    i = 0
    pr = np.exp(-lambda1)
    F=pr
    while ix>=F:
        pr=lambda1/(i+1)*pr
        F = F + pr
        i = i + 1;
    a1 = i
    if a1 == 6:       # Lógica para a letra (a)
        count_a += 1
    if a1 <= 2:       # Lógica para a letra (b)
        count_b += 1

print(f"Probabilidade (a) Exatamente 6: {count_a/N:.4f}")
print(f"Probabilidade (b) No máximo 2: {count_b/N:.4f}")
```

```python
import numpy as np

# LETRA D
q = 0.08
n = 20
N = 100000

c = q / (1 - q)
av = []

for _ in range(N):
    ix = np.random.uniform(0, 1)
    i = 0
    pr = pow((1 - q), n)
    F = pr
    while ix >= F:
        pr = (c * (n - i) / (i + 1)) * pr
        F += pr
        i += 1
    av.append(i)


media_simulada = np.mean(av)
variancia_simulada = np.var(av)

print(f"Média Simulada: {media_simulada:.4f}")
print(f"Variância Simulada: {variancia_simulada:.4f}")
```

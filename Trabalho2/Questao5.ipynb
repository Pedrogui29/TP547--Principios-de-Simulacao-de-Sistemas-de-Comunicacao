```python
import numpy as np
import matplotlib.pyplot as plt


N = 10000
sucessos = 0
estimativas = []
valor_real = 1/6


for i in range(1, N + 1):

    dado1 = np.random.randint(1, 7)
    dado2 = np.random.randint(1, 7)

    soma = dado1 + dado2


    if soma > 9:
        sucessos = sucessos + 1


    estimativas.append(sucessos / i)


prob_final = sucessos / N
print(f"Probabilidade Estimada Final: {prob_final:.6f}")
print(f"Probabilidade Real Teórica: {valor_real:.6f}")

# 4. Gráfico de Convergência (Para ganhar conhecimento)
plt.figure(figsize=(10, 5))
plt.plot(estimativas, label='Estimativa de Monte Carlo')
plt.axhline(y=valor_real, color='r', linestyle='--', label='Valor Real (1/6)')

plt.title('Evolução da Estimativa conforme o N aumenta')
plt.xlabel('Número de Lançamentos (Tentativas)')
plt.ylabel('Probabilidade Calculada')
plt.legend()
plt.grid(True)
plt.show()
```

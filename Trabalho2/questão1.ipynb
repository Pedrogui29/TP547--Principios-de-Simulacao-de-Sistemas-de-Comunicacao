```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
I_real = 2 * np.log(2) - 1
Ns =[10**2, 10**3, 10**4, 10**5]
erros = []

for N in Ns:

    u = np.random.uniform(0, 1, N)


    I_estimado = np.mean(np.log(1 + u))


    erros.append(abs(I_estimado - I_real))
    print(f"N={N}: Estimativa={I_estimado:.6f}, Erro={erros[-1]:.6f}")


plt.figure(figsize=(8, 5))
plt.plot(Ns, erros, 'o-', linewidth=2)
plt.xscale('log')
plt.yscale('log')
plt.title('Convergência de Monte Carlo: Erro Absoluto vs N')
plt.xlabel('Número de Amostras (N)')
plt.ylabel('Erro Absoluto')
plt.grid(True, which="both", ls="-")
plt.show()
```

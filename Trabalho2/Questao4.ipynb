```python
import numpy as np

N = 100000

caixa = np.array([0]*12 + [1]*12 + [2]*12 + [3]*12)


sucessos_sem = 0
for i in range(N):

    mao = np.random.choice(caixa, 7, replace=False)

    if (1 in mao) and (3 in mao):
        sucessos_sem = sucessos_sem + 1

prob_sem = sucessos_sem / N
print("Probabilidade Sem Reposição:", prob_sem)


sucessos_com = 0
for i in range(N):

    mao = np.random.choice(caixa, 7, replace=True)

    if (1 in mao) and (3 in mao):
        sucessos_com = sucessos_com + 1

prob_com = sucessos_com / N
print("Probabilidade Com Reposição:", prob_com)
```

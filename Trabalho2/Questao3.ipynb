```python
import numpy as np
import matplotlib.pyplot as plt

N = 100000
# --- MÉTODO 1: ACEITAÇÃO-REJEIÇÃO ---
amostras_ar = []
tentativas = 0
c = 3 # f(x) maximo em ocorre em x=1, f(1)=3. g(x)=1. c=3/1 

while len(amostras_ar) < N:
    tentativas += 1
    y = np.random.uniform(0, 1)
    u2 = np.random.uniform(0, 1)
    if u2 <= y**2: # Criterio f(y)/(c*g(y)) -> 3y^2 / (3*1) = y^2 
        amostras_ar.append(y)

# --- MÉTODO 2: TRANSFORMADA INVERSA ---
u = np.random.uniform(0, 1, N)
amostras_inv = u**(1/3) # X = U^(1/3) 

# --- COMPARAÇÃO E PLOTAGEM ---
x_teorico = np.linspace(0, 1, 1000)
y_teorico = 3 * x_teorico**2

plt.figure(figsize=(12, 5))

# Plot Aceitação-Rejeição
plt.subplot(1, 2, 1)
plt.hist(amostras_ar, bins=100, density=True, color='blue', alpha=0.6, label='Simulado (A-R)')
plt.plot(x_teorico, y_teorico, 'r', linewidth=2, label='Teórico f(x)')
plt.title(f'Aceitação-Rejeição\nTaxa de Aceitação: {N/tentativas:.2%}')
plt.legend()

# Plot Inversa
plt.subplot(1, 2, 2)
plt.hist(amostras_inv, bins=100, density=True, color='green', alpha=0.6, label='Simulado (Inversa)')
plt.plot(x_teorico, y_teorico, 'r', linewidth=2, label='Teórico f(x)')
plt.title('Transformada Inversa\nEficiência: 100%')
plt.legend()

plt.show()
```
